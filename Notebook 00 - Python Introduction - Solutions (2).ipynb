Note how the above Task 4.3 could also have been solved by writing a `for`-loop that loops over the different rows of `mat1` and in each iteration calls the `np.sort` function. This is however computationally inefficient compared to the *vectorized* method of specifying the `axis`argument as above. **If possible, try to avoid for-loops and try to achieve your tasks by calling a function just once**. This is called vectorized computation. 

We will demonstrate the difference between vectorized and non-vectorized computations in the following. Suppose that, instead of a single 5x5 matrix, you have a set of one million 5x5 matrices. These can be stored in a 3-dimensional numpy array of shape 1000000x5x5.

**Task 6.4**: Write a Python function `bond_value(V0, r, n, M, c)` that computes and returns the capital $Vn$ if an interest of $r > 0$ has been paid on the initial endowment $V0 > 0$ for $n \in \mathbb{N}$ years. If $c = 1$, the variable $r$ refers to a continuously compounded rate of return (i.e. $Vn = V0 \exp(rn)$), and if $c = 0$, it refers to a simple rate paid over $M$ time periods per year (i.e. $Vn = V0(1 + \frac{r}{M})^{n\cdot M}$). Your function should be able to work in a vectorized manner, which means that one should be able to provide the arguments `V0, r, n, M, c` as arrays with multiple values. In this case, the output of the function should also be an array of suitable size. Also, define a default value of your choice for the argument $M$ so that the user does not need to specify it when he uses continuous compounding.

```python
def bond_value(V0, r, n, c, M=1):
    if c == 0:
        return V0 * (1 + r / M) ** (n * M)
    elif c == 1:
        return V0 * np.exp(r * n)
    else:
        raise Exception('Error: Values for c have to be 0 or 1')
```

**Task 6.5**: Evaluate your function at the inputs `V0 = 1000`, `r = 0.025`, `n = np.arange(1, 51)` and `c = 1` and store the result in an array `capital_cont_rate`. Then, evaluate your function at the inputs `V0 = 1000`, `r = 0.025`, `n = np.arange(1, 51)`, `c = 0` and `M = 2` and store the result in an array `capital_simple_rate`.

```python
V0 = 1000
r = 0.025
n = np.arange(1, 51)
print(n)
capital_cont_rate = bond_value(V0, r, n, 1)
capital_simple_rate = bond_value(V0, r, n, 0, M = 2)

print(capital_cont_rate)
print(capital_simple_rate)
```

## Topic 7: Debugging and dealing with error messages

Often during programming you will face error messages because somewhere in your code there is still some mistake. In this case you need to be able to understand what the error message is trying to tell you. The following snippets each contain some erroneous code that produces an error. When dealing with errors, you generally have two options: If it is an error that occurs in every circumstance because your code is wrong, you have to fix it. But it can also happen that there are some errors that occur inside a for-loop only in some iterations of the loop or inside a function only for some input values. In this case you can catch the error inside a `try`-`except`-block as follows:

<code> 
try: 
    some_code_that_can_throw_an_error
except [ErrorType]:
    do_the_following 
</code>
<br><br>
The `do_the_following` code inside of the `except` block is only executed if the code `some_code_that_can_throw_an_error` inside the `try` block throws an error. If the code inside the `try` block does not throw an error, it is executed normally. You also do not have the specify the `ErrorType` in the `except` statement.

**Task 7.8**: The following code snippet is supposed to compute the function value $z = f(x, y) = 1 + \frac{3}{\exp(-x) + \exp(-y)}$ for some inputs $x$ and $y$. After defining some values for $x$ and $y$, it is supposed to compute the corresponding values $z = f(x, y)$ in a for-loop. However, during computation it throws a `ZeroDivisionError`. Wrap the body of the function inside a `try`-`except`-block and let the function return the value `np.inf` whenever it throws a `ZeroDivisionError`.

```python
import math

def f(x, y):
    return 1 + 3 / (math.exp(-x) + math.exp(-y))

x = np.arange(2000)
y = x / 2
z = np.zeros(2000)


for i in range(2000):
    try:
        z[i] = f(x[i], y[i])
    except ZeroDivisionError:
        z[i] = np.inf

print(x)
print(y)
print(z)
```
